```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [ "LangChain is a framework for building context-aware reasoning applications", 
             "Vector embeddings convert text into numerical representations that capture semantic meaning",
               "Texts with similar meanings are placed closer together in the vector space",
                 "Embeddings enable efficient similarity search and retrieval from large collections of data",
                   "They are a fundamental component of Retrieval-Augmented Generation applications" 
                   ]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview")
vector = embeddings.embed_documents(documents)
```

```python
len(vector[0])
```

#### Store Vector Embeddings in Vector DB

```python
from langchain_chroma import Chroma
```

```python
from langchain_chroma import Chroma

Vector_store = Chroma.from_texts(
    texts=documents,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query = "semantic meaning"
result = Vector_store.similarity_search(query=query ,  k=2)
result
```
