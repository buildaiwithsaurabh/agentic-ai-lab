```python
from dotenv import load_dotenv
load_dotenv()

```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number(a: int , b:int) -> int:
    """
         Add two numbers together.
    
         Args:
           a : Number One
           b : Number Two
    """
    return a + b
```

```python
add_number.invoke({"a": 12, "b": 13})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model = "openai/gpt-oss-20b")
```

```python
agent = create_agent(
  model=llm,
  tools=[add_number],
  system_prompt="Your are math teacher for calculation")
```

```python
res = agent.invoke({"messages": [{"role": "user", "content": "What is 12 + 13?"}]})
```

```python
# If res is a State object containing a list of messages:
print(res["messages"][-1].content)
```
